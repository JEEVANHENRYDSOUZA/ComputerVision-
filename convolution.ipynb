```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
from scipy.signal import convolve2d
path = './images.jpeg'
input_image=cv.imread(path)
sobel_filter = np.array([[1, 2, 1], [0, 0, 0], [-1, -2, -1]])
output1 = cv.filter2D(input_image, -1, sobel_filter)
## the second parameter is the depth of the output image, -1 means same as input image
## it is ddepth not image depth , making sure the data type is within bounds
        
gray = cv.cvtColor(input_image, cv.COLOR_BGR2GRAY)
output = convolve2d(gray, sobel_filter, mode='valid')
### this will need a 2d image to apply a 2d filter 
### our input image is a 3d image with 3 channels (RGB) so we need to convert it to grayscale first

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(input_image)
plt.colorbar()
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(output1)
plt.title("Convolved Image x edge detection")
plt.colorbar()

plt.axis("off")


### subplot we metion the rows and columns for the first two para the last is the index number of our image
plt.subplot(1, 3, 3)
plt.imshow(output)
plt.title("Convolved Image using scipy mode is full no padding")
plt.axis("off")
plt.show()

print(f"The shape of the input image is: {input_image.shape}")
print(f"The shape of the output1 image is: {output1.shape}")
print(f"The shape of the gray image is: {gray.shape}")
print(f"The shape of the output image is: {output.shape}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
from scipy.signal import convolve2d
#### every image is a 2d array of pixels, each pixel has a value between 0 and 255 for grayscale images and 3 values for RGB images
numpy_array=np.zeros((32,32))
print(numpy_array)
## the array of 32*32 is created to represent a grayscale image of 32*32 pixels, each pixel has a value of 0 which means black color
## inserting a unit impulse in the center of the image
numpy_array[16,16]=1
print(numpy_array[16,16])
sobel_filter = np.array([[1, 2, 1], [0, 0, 0], [-1, -2, -1]])
output = convolve2d(numpy_array, sobel_filter, mode='full')

plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(numpy_array, cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.colorbar()

plt.subplot(1, 3, 2)
plt.imshow(sobel_filter, cmap='gray')
plt.title("Sobel Filter")
plt.axis("off")
plt.colorbar() 

plt.subplot(1, 3, 3)
plt.imshow(output, cmap='gray')
plt.title("Convolved Image using scipy mode is full no padding")
plt.axis("off")
plt.show()
```
